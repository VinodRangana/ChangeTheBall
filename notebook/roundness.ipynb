```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import math

# 1. Load your test image (Update the filename if yours is different)
image = cv2.imread('interim/image.png')

# 2. Pre-processing for the Black & White test image
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# 3. Find the contour
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

if contours:
    # Get the largest contour (the circle itself)
    largest_contour = max(contours, key=cv2.contourArea)
    
    # Create a copy of the image to draw on
    output_img = image.copy()
    
    # Draw the original contour detected by OpenCV in RED
    cv2.drawContours(output_img, [largest_contour], -1, (255, 0, 0), 3)

    # --- METRIC 1: CIRCULARITY ---
    area = cv2.contourArea(largest_contour)
    perimeter = cv2.arcLength(largest_contour, True)

    if perimeter > 0:
        circularity = (4 * np.pi * area) / (perimeter ** 2)
    else:
        circularity = 0.0

    # --- METRIC 2: ECCENTRICITY (ELLIPSE FIT) ---
    eccentricity = 0.0
    if len(largest_contour) >= 5: # fitEllipse requires at least 5 points
        ellipse = cv2.fitEllipse(largest_contour)
        (center_x, center_y), (minor_axis, major_axis), angle = ellipse
        
        eccentricity = minor_axis / major_axis
        
        # Draw the perfectly calculated mathematical ellipse in GREEN
        cv2.ellipse(output_img, ellipse, (0, 255, 0), thickness=2)

    # --- PRINT RESULTS ---
    print("--- Sphericity Metrics ---")
    print(f"perimeter: {perimeter:.5f}")
    print(f"area: {area:.5f}")
    print(f"Circularity (Area/Perimeter): {circularity:.5f}")
    print(f"Eccentricity (Ellipse Ratio): {eccentricity:.5f}")

    # --- VISUALIZE ---
    # Convert BGR to RGB for correct matplotlib color display
    output_img_rgb = cv2.cvtColor(output_img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(6, 6))
    plt.title(f"Circ: {circularity:.4f} | Ecc: {eccentricity:.4f}")
    plt.imshow(output_img_rgb)
    plt.axis('off')
    plt.show()
else:
    print("No contours found in the image.")
```
